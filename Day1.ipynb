 # Band Name Generator

```python
print("Good day to you! Welcome to the Band Name Generator.")

city = input("Which city did you grow up in?\n ")
pet = input("What was your pet's name?\n")
food = input("What was your favorite food?\n")

band_name = city + " "+ pet+ " "+ food

print("Your band name is",band_name.title())
print ("Your band name could be",city.upper(), pet.upper(), food.upper())
```
