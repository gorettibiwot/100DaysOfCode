```python
def add(n1, n2):
    return n1 + n2
    
def subtract(n1, n2):
    return n1 - n2
    
def multiply(n1, n2):
    return n1 * n2
    
def division(n1,n2):
    if n2 == 0:
        return None
    return n1/n2

def exponent(n1,n2):
    return n1 ** n2

def modulus(n1, n2):
    if n2 == 0:
        return None
    return n1 % n2

operations = {
    '+': add,
    '-': subtract,
    '*': multiply,
    '/': division,
    '**': exponent,
    '%': modulus
}

def calculator():
    num1 = float(input('Enter first number:'))
    should_continue = True
    
    while should_continue:
        operand = input('Choose operand [+,-,*,/,**,%]:')
        if operand not in operations:
            print('Invalid input, try again')
            continue
    
        num2 = float(input('Enter next number:'))
        answer = operations[operand](num1, num2)
                
        if answer is None:
            print('Cannot divide by zero')
            continue

        print(f'{num1} {operand} {num2} = {answer}')
        
        another = input('Do you want to continue (y/n):').lower().strip()
        if another == 'y':
            num1 = answer
        else: 
            should_continue = False

while True:
    calculator()
    again = input("Type 'q' to quit, or press Enter to start over:")
    if again == 'q':
        break
        
```
