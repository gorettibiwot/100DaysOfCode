```python
        
print ('You will be prompted to enter the lengths of sides of a tringle separately')
     
a= float(input(' Side a: '))
b= float(input(' Side b: '))
c= float(input(' Side c: '))

if a > 0 and b > 0 and c > 0 and a + b > c and a + c > b and b + c > a:
    
    if a == b == c:
       print('The triangle is equilateral')
        
    elif a == b or a == c or b ==c:
       print(' The triangle is isosceles')

    else:
        print('Scalene triangle')
        
else:
    print('These sides cannot form a triangle')
```

```python

username = input('Please enter username').strip()
errors = 0

if (len(username) < 5):
    print('Username is too short')
    errors += 1
    
if (len(username) >15):
    print('Username is too long')
    errors += 1

if ' ' in username:
     print('Username cannot have spaces')
     errors += 1

if not username.isalnum():
     print('Username can only contain letters and digits')
     errors += 1

if username != '' and username[0].isdigit():
    print('Username cannot start with a digit')
    errors += 1

if errors == 0:
    print('Username is valid')
```

```python
phone = input('Enter phone number:').strip().replace(' ','').replace('-','')

if not phone.isnumeric():
    print('Phone number can only contain digits, spaces and dashes')
elif len(phone) != 10:
    print('Phone number must have exactly 10 digits')
elif phone[0] != '0':
    print('Phone number must start with 0')
else:
    print(f'Phone number is valid: {phone}')
```

```python
pin= 1234
balance = 500

enter_pin = int(input('Please enter pin:'))

if enter_pin == pin:
    withdraw = int(input('How much would you like to withdraw?'))
    if withdraw <= 0:
        print ('Invalid amount')
    elif withdraw > balance:
        print ('Insufficient funds.')
    else:
        print(f'Your balance is {(balance - withdraw)}')
else:
    print('Wrong pin')
    
```
