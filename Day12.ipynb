# Scope and Guessing game

```python
import random

EASY_LEVEL_TURNS = 10
HARD_LEVEL_TURNS = 5

def set_difficulty():
    difficulty = input('Choose a difficulty. Type "easy" or "hard":').lower().strip()
    if difficulty == 'easy':
        return EASY_LEVEL_TURNS
    else:
        return HARD_LEVEL_TURNS
    


def check_answer(guess, computer_pick, turns):
    if guess > computer_pick:
        print('Too high')
        return turns - 1
        
    elif guess < computer_pick:
        print('Too low')
        return turns - 1

    else:
        print(f'You got it! The answer was {computer_pick}.')
        return turns

def game():

    print('Welcome to the Number Guessing Game!')
    print('I am thinking of a number between 1 and 100')
    
    computer_pick = random.randint(1,100)
    turns = set_difficulty()
    guess = 0

    while guess != computer_pick:
        print(f'You have {turns} attempts remaining to guess the number')
        guess = int(input('Make a guess:'))
        turns = check_answer(guess, computer_pick, turns)

        if turns == 0:
            print(f'You have run out of guesses. The number was {computer_pick}. You lose')
            return
        elif guess != computer_pick:
            print('Guess again')

game()
```
