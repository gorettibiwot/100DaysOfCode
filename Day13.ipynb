```python
def average(numbers):
    total = 0
    for n in numbers:
        total += n
    return total / len(numbers)

print(average([2, 4, 6]))
```

```python
def add_card(hand, card):
    hand.append(card)
    return sum(hand)

hand_score = add_card([5, 6], 10)
print(hand_score + 1)
```

```python
password = 'abc def'
errors = 0

if len(password) < 8:
    print('Too short')
    errors += 1
if ' ' in password:
    print('Spaces not allowed')
    errors += 1
if errors == 0:
    print('Valid')
```
