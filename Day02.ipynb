```python
#tip calculator 

bill=float(input('What is the bill?\n'))
tip_percentage = int(input('What tip percentage would you like to give?\n'))
people= int(input('How many people are contributing?\n'))

tip = bill*tip_percentage / 100
total = bill + tip
topay= total/people


print(f'The bill is Ksh {bill:.2f}, plus a {tip_percentage}% tip of Ksh {tip:.2f}, for a total of Ksh {total:.2f}.')
print(f'Each person will contribute Ksh {topay}.')
        

```

```python
value = '968-Maria, ( D@t@ Engineer );; 27y  '

clean = value.strip().lower().replace('@', 'a')
clean = clean[4:]                

name, rest = clean.split(',')    
role, age = rest.split(';;')     

role = role.replace('(', '').replace(')', '').strip()   
age = age.strip().replace('y', '')                      

print(f"name: {name} | role: {role} | age: {age}")
```

```python
challenge= '  ID#0042 | PRICE: $19.99 | QTY: 3  '
clean= challenge.strip()

ID, price, qty= clean.split('|')

item_id = ID.replace('#',' = ')
item_price = float(price.split('$')[1])
item_qty= int(qty.split(':')[1])

total = item_price * item_qty


print(f'{item_id}| Price = {item_price}| QTY:{item_qty}')
print(f'Total cost: {round(total,2)}')
```
