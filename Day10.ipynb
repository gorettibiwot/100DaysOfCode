```python
def add(n1, n2):
    return n1 + n2

def subtract(n1,n2):
    return n1 - n2

def multiply(n1,n2):
    return n1 * n2

def divide(n1, n2):
    if n2 == 0:
        return None
    return n1/n2

operations = {
    '+': add, 
    '-': subtract,
    '*': multiply, 
    '/': divide
}




n1 = float(input('Enter first number:'))
should_continue = True

while should_continue:
    n2 = float(input('Enter the next number:'))
    user_pick = input('Enter symbol[+, -, /,*]:')
    for symbol, action in operations.items():
        if user_pick == symbol:
            answer = action(n1,n2)
            print(answer)
        
       
    another =  input('Do you want to continue? y/n')
    if another == 'y':
        n1 = answer

    else:
        should_continue = False

         
```

```python
prices = {'milk': 2.5, 'eggs': 4.0, 'bread': 1.8}
total = 0


def expensive_item(prices):
    total = 0
    highest_price = 0
    highest_item = ''
    for item, amount in prices.items():
        total += amount
        if amount > highest_price:           
            highest_price = amount
            highest_item = item
            
    return total, highest_item, highest_price
        

total, item, amount  = expensive_item(prices)
print(f'The total is {round(total,2)} and the most expensive item was {item} (${amount})')
    
```
