# Ceasar cipher 

```python
alphabet = 'abcdefghijklmnopqrstuvwxyz'

def shift_letter(letter,shift):
    x = len(alphabet)
    position = alphabet.index(letter)
    new_position = (position + shift) % x
    return alphabet[new_position]

def shift_word(word,shift):
    new_word = ''
    for letter in word.lower():
        if letter in alphabet:    
            new_word += shift_letter(letter, shift)
        else:
            new_word += letter
    return new_word

def ceaser(text, shift, direction):
    if direction == 'decode':
        shift = -shift
    return shift_word(text, shift)
           
            
direction = input('Type "encode" to encrypt and "decode" to decrypt:').lower()
text = input('Type your message here:').lower()
shift = int(input('Type the shift number:'))


if direction == 'encode' or direction == 'decode':
    print(f'Here is the {direction} result:{ceaser(text,shift,direction)}')
else:
    print('Invalid Input')
```
