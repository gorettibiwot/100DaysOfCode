# Blackjack Capstone

```python
import random 
        
deck = [11, 2, 3, 4, 5, 6, 7, 8, 9, 10, 10, 10, 10]
        
def deal_cards():
    return random.choice(deck)

def score_keeping(cards):
    if sum(cards) == 21 and len(cards) == 2:
        return 0
    if sum(cards) > 21 and 11 in cards:
        cards[cards.index(11)] = 1
    return sum(cards)
       
        
def compare(player_score, computer_score):
    if player_score == computer_score:
        return 'Draw'
    elif player_score == 0:
        return 'Blackjack, You win!'
    elif computer_score == 0:
        return 'You lose, opponent has Blackjack'
    elif player_score > 21:
        return 'Bust. You went over'
    elif computer_score > 21:
        return 'You win, Opponent went over'
    elif player_score > computer_score:
        return 'You win'
    else:
        return 'You lose'

def play_game():
    player_cards = []
    computer_cards = []
    
    for _ in range(2):
        player_cards.append(deal_cards())
        computer_cards.append(deal_cards())

    player_score = score_keeping(player_cards)
    computer_score = score_keeping(computer_cards)
    is_game_over = False
        
    print(f'Your cards: {player_cards}, current score: {player_score}')
    print(f"Computer's first card: {computer_cards[0]}")

        #player
    while not is_game_over:
        if player_score == 0 or computer_score == 0 or player_score > 21:
            is_game_over = True
        else:
             another_round = input("Type 'y' to pick a card, 'n' to pass:").lower().strip()
             if another_round == 'y':
                 player_cards.append(deal_cards())
                 player_score = score_keeping(player_cards)
                 print(f'Your cards: {player_cards}, current score: {player_score}')
             else:
                 is_game_over = True
        
    #computer
    while computer_score != 0 and computer_score < 17:
        computer_cards.append(deal_cards())
        computer_score = score_keeping(computer_cards)
       
    
    print(f"Your final hand: {player_cards}, final score: {player_score}")
    print(f"Computer's final hand: {computer_cards}, final score: {computer_score}")
    print(compare(player_score, computer_score))

continue_playing = True
while continue_playing:
    play_game()
    another_game = input('Do you want to play another game? [y/n]:').lower().strip()
    if another_game != 'y':
        continue_playing = False
    
    
```
