```python
email = ''
email =  email.strip()

if email == '':
    print('Email cannot be empty')
    
elif not('.' in email and '@' in email):
    print('Email must contain . and @)')

elif email.count('@') != 1:
    print('Email must contain exactly one @')

elif not email.endswith(('.com', '.org', '.net')):
    print('Email must end with .com, .org or .net')

elif len(email) >254:
    print('Email must not be longer than 254 characters')

elif not (email[0].isalum() and email[-1].isalum()):
    print ('Email must start and end with a letter or digit')      

else:
    print('Email is valid')


```

```python
#validate the quality and correctness of passwords
email= input('Enter email:')
email = email.strip()

password = input('Enter password:')
password = password.strip()

is_valid = False

if password == '':
    print ('password cannot be empty')
    is_valid = False


if not any(c.isupper() for c in password ):
    print('Password must contain at least one uppercase')
    is_valid = False
    

if not any (c.islower() for c in password):
    print('Password must contain at least one lowercase letter')
    is_valid = False


if password.lower() == email.lower() :
    print ('Password cannot be the same as the email.')
    is_valid = False



if ' ' in password:
    print('Password cannot have any spaces')
    is_valid = False 


if not (password[0].isalnum() and password[-1].isalnum()):
    print('Password must start and end with a digit or letter')
    is_valid = False
    
if is_valid:
    print('Password is valid')

```
