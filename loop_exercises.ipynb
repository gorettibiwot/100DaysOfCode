```python
scores = [72, 85, 90, 64, 58]
total = 0
count = 0
count_70 = 0
highest = scores[0]

for score in scores:
    total += score
    count += 1
    if score >= 70:
        count_70 += 1
    if score > highest:
        highest = score

print(f'Average: {total / count:.1f}')     
print(f'Scores of 70 or above: {count_70}') 
print(f'Highest: {highest}')               
    
```
