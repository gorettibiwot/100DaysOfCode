```python
bill = float(input('What is the bill?'))
percent = int(input('What percent would you like to tip?'))
people = int(input('How many are sharing costs?'))


def share_per_person(bill, percent, people):
    tip = (bill * percent)/ 100
    amount = (bill + tip) / people
    return round(amount, 2)



print(f'The bill was {bill}, adding {percent}% tip, with {people} people sharing')
print(f'The total amount for each comes to {share_per_person(bill,percent,people)}')
```

```python
def grade(score):
    if score >= 90:
        return 'A'
    if score >= 80:
        return 'B'
    if score >= 70:
        return 'C'
    if score >= 60:
        return 'D'
    if score >= 50:
        return 'E'
    return 'Fail'


    
x= int(input('Score'))
print(f'Your final score is {x} and grade is {grade(x)}')
```

```python
password= input('Enter password:')

def is_valid(password):
    for p in password:
        if len(password) < 5:
           print ('Password too short')
           return False
        if len(password) > 20:
            print ('Password too long')
            return False
    
        has_upper = False
        has_lower = False
        for ch in password:
            if ch.isupper():
                has_upper = True
            if ch.islower():
                has_lower = True
    
        if not has_upper:
            print('Password must contain at least one uppercase letter')
            return False
        if not has_lower:
            print('Password must contain at least one lowercase letter')
            return False
        if ' ' in password:
            print('Password cannot contain spaces')
            return False
    
        return True


is_valid(password)
```

```python
sentence = input('Enter sentence').lower()

def count_vowels(sentence):
    vowels = ('aeiou')
    count = 0
    for s in sentence:
        if s in vowels:
            count += 1
    return count
        

count_vowels(sentence)
```
