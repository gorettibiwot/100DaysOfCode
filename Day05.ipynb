```python
import random

alpha = list('abcdefghijklmnopqrstuvwxyzABCDEFGHIJKLMNOPQRSTUVWXYZ')
numer = list('0123456789')
sym = list('!@#$%^&*~+-?')

print('Welcome to the password generator')
nr_alpha = int(input('How many letters would you like?:'))
nr_numer = int(input('How many numbers would you like?:'))
nr_sym= int(input('How many symbols would you like?:'))


password2 = []
for _ in range(nr_alpha):
    password2.append(random.choice(alpha))

for _ in range(nr_numer):
    password2.append(random.choice(alpha))

for _ in range(nr_sym):
    password2.append(random.choice(sym))

random.shuffle(password2)
hard_password = ''.join(password2)

print(f'Your password is {hard_password}')

                    
```

```python
numbers = input('Enter numbers or "done":').lower()
total = 0
count = 0 

while numbers != 'done':
    total += float(numbers)
    count += 1
    numbers = input('Enter numbers or "done":').lower()
    

if count > 0:
    print(f'The sum is {total}')
    print(f'Average is {total / count:.2f}')
else:
    print('No numbers entered')
```

```python
import random

guess_count = 0
computer = random.randint(1,100)

guess = int(input('Guess number:'))
guess_count += 1

while guess != computer:
    if computer > guess:
        print('Higher')
    else:
        print('Lower')
    guess = int(input('Guess again:'))
    guess_count += 1


print(f'Correct! You got it in {guess_count} guesses.')
```

```python
#Triangles: print a right triangle of stars with 5 rows, then an upside-down version (5 stars down to 1).
star = '*'

for i in range(1, 6):
    print(i * star)

for i in range(5, 0, -1):
    print(i * star)

for row in range(1, 6):
    for _ in range(row):
        print('*', end='')
    print()
for row in range(5, 0, -1):
    for _ in range(row):
        print('*', end='')
    print()
```
